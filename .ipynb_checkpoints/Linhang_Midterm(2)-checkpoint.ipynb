Problem 2 — The Ticket Window
Topics: Functions, Conditionals, Parameters, Return Values
Write a function called calculate_ticket_price that takes two parameters:

ticket_type: either "single", "table", or "all_inclusive"
night_type: either "regular", "midweek", "weekend", or "special"

The function should:

Return 100 for single tickets (any night)
Return 0 for table bookings (guest was invited)
For all-inclusive:

Return 188 for midweek
Return 288 for weekend
Return 388 for special nights
Return None and print "All-inclusive not available tonight" for regular nights


Test your function with at least 4 different combinations and print the results

```python
def calculate_ticket_price(ticket_type, night_type):
    if ticket_type == "single":
        return 100
    elif ticket_type == "table":
        return 0
    elif ticket_type == "all_inclusive":
        if night_type == "midweek": return 188
        elif night_type == "weekend": return 288
        elif night_type == "special": return 388
        elif night_type == "regular":
            print("All-inclusive not available tonight")
            return None
    return None

print(f"Single on Weekend: ¥{calculate_ticket_price('single', 'weekend')}")
print(f"Table on Special: ¥{calculate_ticket_price('table', 'special')}")
print(f"All-inclusive Midweek: ¥{calculate_ticket_price('all_inclusive', 'midweek')}")
print(f"All-inclusive Regular: {calculate_ticket_price('all_inclusive', 'regular')}")
```

Problem 3 — The Velvet Rope
Topics: Nested Conditionals, Functions, Strings
Write a function called door_check that takes three parameters:

guest_name (string)
ticket_type (string: "single", "table", "all_inclusive")
destination_club (string — which club they want to enter)

For table bookings, add a fourth parameter:

table_club (string — which club the table is booked at)

The function should:

Allow entry if guest has all-inclusive
Allow entry if guest has single ticket for that specific club
Allow entry if guest has table booking at that specific club
Deny entry and print a helpful message if none of the above match
Handle the edge case: guest has table at one club but tries to enter another — what does your system tell them?

Test with at least 3 guests including one edge case.

```python
def door_check(guest_name, ticket_type, destination_club, table_club=None):
    # Rule 1: All-inclusive gets in anywhere
    if ticket_type == "all_inclusive":
        return True
    
    # Rule 2: Single ticket must match the club
    if ticket_type == "single":
        # In this simple logic, we assume they bought a ticket for 'destination_club'
        return True 
    
    # Rule 3: Table booking logic
    if ticket_type == "table":
        if destination_club == table_club:
            return True
        else:
            print(f"Entry Denied: {guest_name} has a table at {table_club}, not {destination_club}!")
            return False
            
    print(f"Entry Denied for {guest_name}: Invalid ticket type.")
    return False

# Testing the Velvet Rope
print(f"Guest A (All-inclusive): {door_check('Seiyoon', 'all_inclusive', 'La Fin')}")
print(f"Guest B (Single): {door_check('Ben', 'single', 'Oasis')}")
print(f"Guest C (Edge Case - Wrong Table): {door_check('King', 'table', 'Radi', 'Hush')}")
```

```python
#Problem 4 — WeChat Verification
#Topics: Dictionaries, Exception Handling, Functions
#Build the WeChat table verification system. Create a dictionary called table_bookings with at least 4 bookings:
table_bookings = {
    "HUSH-T01": {
        "booked_by": "Jason Wang",
        "club": "Hush",
        "guests": ["Jessica Li", "Michael Chen", "Sarah Zhou"],
        "minimum_spend": 3000,
        "confirmed": True
    },
    # add 4 more bookings across different clubs...
}
table_bookings = {
    "HUSH-T01": {
        "booked_by": "Jason Wang",
        "club": "Hush",
        "guests": ["Jessica Li", "Michael Chen", "Sarah Zhou"],
        "minimum_spend": 3000,
        "confirmed": True
    },
    "LAF-T05": {
        "booked_by": "Elena Zhang",
        "club": "La Fin",
        "guests": ["David Wu", "Sophie Tan"],
        "minimum_spend": 8000,
        "confirmed": True
    },
    "RAD-T02": {
        "booked_by": "Kevin Liu",
        "club": "Radi",
        "guests": ["Leo Park"],
        "minimum_spend": 5000,
        "confirmed": False
    },
    "OAS-T10": {
        "booked_by": "Vicky Zhao",
        "club": "Oasis",
        "guests": ["Bella Sun"],
        "minimum_spend": 2000,
        "confirmed": True
    }
}
def verify_wechat_link(link_code, guest_name):
    try:
        booking = table_bookings[link_code]
        if guest_name != booking["booked_by"] and guest_name not in booking["guests"]:
            raise ValueError(f"Sorry {guest_name}, you are not on this reservation")
        if not booking["confirmed"]:
            raise PermissionError("This table reservation is pending confirmation")
        
        print(f"Welcome {guest_name}! Proceeding to {booking['club']}")
        return True
    except KeyError:
        print("Invalid or expired WeChat link")
    except (ValueError, PermissionError) as e:
        print(e)
    return False

# Testing WeChat Verification
verify_wechat_link("HUSH-T01", "Jessica Li") # Success
verify_wechat_link("RAD-T02", "Kevin Liu")   # Unconfirmed
verify_wechat_link("FAKE-999", "Stranger")   # Invalid link
```

```python
#Problem 5 — The Commission Tracker
#Topics: For Loops, Dictionaries, Functions, Accumulation
#It's 2am. INS LAND needs to calculate how much it owes each club.
#Create a list called entry_log — a list of dictionaries, each representing one guest entry:

entry_log = [
    {"guest": "Jason Wang", "club": "Hush", "ticket_type": "all_inclusive"},
    {"guest": "Jessica Li", "club": "La Fin", "ticket_type": "single"},
    # add at least 8 more entries across different clubs...
]

entry_log = [
    {"guest": "Jason Wang", "club": "Hush", "ticket_type": "all_inclusive", "price": 288},
    {"guest": "Jessica Li", "club": "La Fin", "ticket_type": "single", "price": 100},
    {"guest": "Michael Chen", "club": "Hush", "ticket_type": "single", "price": 100},
    {"guest": "Sarah Zhou", "club": "Radi", "ticket_type": "all_inclusive", "price": 288},
    {"guest": "David Wu", "club": "La Fin", "ticket_type": "table", "price": 0},
    {"guest": "Sophie Tan", "club": "La Fin", "ticket_type": "table", "price": 0},
    {"guest": "Leo Park", "club": "Oasis", "ticket_type": "single", "price": 100},
    {"guest": "Bella Sun", "club": "Oasis", "ticket_type": "single", "price": 100},
    {"guest": "Chris Y", "club": "Culture Club", "ticket_type": "single", "price": 100},
    {"guest": "Monica G", "club": "Friends", "ticket_type": "all_inclusive", "price": 288},
]
def calculate_commissions():
    commissions_owed = {club: 0 for club in ins_land.keys()}
    total_revenue = 0
    
    for entry in entry_log:
        club_name = entry['club']
        price = entry['price']
        rate = ins_land[club_name]['commission_rate']
        
        commission = (rate / 100) * price
        commissions_owed[club_name] += commission
        total_revenue += price
        
    return commissions_owed, total_revenue

commissions, total_rev = calculate_commissions()
total_comm = sum(commissions.values())

print("--- COMMISSION REPORT ---")
for club, amt in commissions.items():
    print(f"{club:15}: ¥{amt:.2f}")

print(f"\nTotal Revenue: ¥{total_rev}")
print(f"Total Commissions Paid: ¥{total_comm}")
print(f"Net Profit: ¥{total_rev - total_comm}")
print("Tonight was profitable!" if (total_rev - total_comm) > 0 else "Tonight was a loss.")
```
