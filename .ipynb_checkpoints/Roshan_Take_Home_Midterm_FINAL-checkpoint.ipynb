## Question 1: Indentation Matters (10 points)

Let's get the ball rolling. One of the most important features of Python is **indentation**. Indentation determines which lines of code belong to a loop and which lines are executed outside the loop. Even when the **lines of code are identical**, changing the indentation can change how the program behaves and what output it produces.

Below are **three code snippets**. The **only difference between them is the indentation**.

For each snippet:

- Carefully read the code.
- **Explain in words what the program prints and why.**
- In your explanation, focus on **which lines of code are executed inside the loop and which are executed outside the loop**.

````python
CODE # 1

numbers = [2, 4, 6]
total = 0

for n in numbers:
    total = total + n
    total = total + 1

print(total)

CODE # 2

numbers = [2, 4, 6]
total = 0

for n in numbers:
    total = total + n

    total = total + 1

print(total)

CODE # 3

numbers = [2, 4, 6]
total = 0

for n in numbers:
    total = total + n

total = total + 1

print(total)
````

Answer:

Code 1 prints: 15

Both "total = total + n" and "total = total + 1" are inside the loop. Each iteration adds both n and 1 to total.
- Iteration 1: total = 0 + 2 + 1 = 3
- Iteration 2: total = 3 + 4 + 1 = 8
- Iteration 3: total = 8 + 6 + 1 = 15

Code 2 prints: 15

Same as Code 1 — blank lines do not change indentation in Python. Both lines are still inside the loop, so the result is the same: 15.

Code 3 prints: 13

Only "total = total + n" is inside the loop. "total = total + 1" is outside the loop and only runs once after the loop finishes.
- After the loop: total = 0 + 2 + 4 + 6 = 12
- Then add 1 ---> 13

## Question 3: Traversing Data (20 points)

Economists often work with cross-country data to compare living standards and economic performance.
Below is a dataset of 6 countries, each described by four indicators: `gdp_per_capita` (in USD),
`unemployment_rate` (in %), `inflation_rate` (in %), and `life_expectancy` (in years). Run the cell below to load the data.

**(a)** Write a `for` loop that prints each country's name and its `gdp_per_capita` in the following format: *(5 points)*
```
USA: $65000
Germany: $48000
...
```

**(b)** Write a function called `find_max` that takes the `countries` dictionary and the name of an indicator (as a string) as arguments, and returns the name of the country with the **highest value** for that indicator. *(10 points)*

**(c)** Using your `find_max` function, print the country with the highest `life_expectancy` and the country with the highest `inflation_rate`, using f-strings in the following format: *(5 points)*
```
The country with the highest life expectancy is Japan.
The country with the highest inflation rate is Nigeria.
```

````python
countries = {
    "USA":     {"gdp_per_capita": 65000, "unemployment_rate": 3.9, "inflation_rate": 3.2, "life_expectancy": 79},
    "Germany": {"gdp_per_capita": 48000, "unemployment_rate": 3.1, "inflation_rate": 2.8, "life_expectancy": 81},
    "Brazil":  {"gdp_per_capita": 8500,  "unemployment_rate": 9.2, "inflation_rate": 6.1, "life_expectancy": 75},
    "India":   {"gdp_per_capita": 2200,  "unemployment_rate": 7.8, "inflation_rate": 5.5, "life_expectancy": 70},
    "Nigeria": {"gdp_per_capita": 2100,  "unemployment_rate": 5.3, "inflation_rate": 18.6, "life_expectancy": 55},
    "Japan":   {"gdp_per_capita": 42000, "unemployment_rate": 2.6, "inflation_rate": 1.1,  "life_expectancy": 84},
}
````

````python
# Answer:

# Part (a)

for country in countries:
    print(f"{country}: ${countries[country]['gdp_per_capita']}")

# Part (b)

def find_max(countries, indicator):
    best_country = None
    best_value = 0
    for country in countries:
        if countries[country][indicator] > best_value:
            best_value = countries[country][indicator]
            best_country = country
    return best_country

# Part (c)

print(f"The country with the highest life expectancy is {find_max(countries, 'life_expectancy')}.")
print(f"The country with the highest inflation rate is {find_max(countries, 'inflation_rate')}.")
````

## Question 4: Finding the Most Similar Individual (40 points)

In economics, we often evaluate the effect of a policy or intervention by comparing "treated" individuals (individuals that experienced the policy/intervention) to untreated individuals who are as similar as possible. Hence, it can be very useful in evaluating policy to work with large scale data to find each individual's "closest match". 

Below is a dataset of 5 individuals, each described by three numeric characteristics: `age`, `income` (in thousands), and `years_of_education`. Run the cell below to load the data.

**Your task:**

**(a)** Write a function called `total_distance` that takes two names (strings) and the `individuals` dictionary as arguments, and returns the **total absolute difference** across all three characteristics between those two individuals. *(10 points)*

**(b)** Write a function called `find_match` that takes a name and the `individuals` dictionary as arguments, and returns the name of the **most similar** individual — i.e., the one with the smallest total distance. The function should **not** compare the individual to themselves. *(25 points)*

**(c)** Use your `find_match` function to find the best match for **"Bob"** and print the result using an f-string in the following format: *(5 points)*
```
The best match for Bob is Carlos with a total distance of 17.
```

> **Hint:** To find the most similar individual, you may find it useful to keep track of the smallest distance seen so far as you loop through the dictionary.

````python
individuals = {
    "Alice":  {"age": 28, "income": 45, "years_of_education": 14},
    "Bob":    {"age": 35, "income": 62, "years_of_education": 16},
    "Carlos": {"age": 30, "income": 48, "years_of_education": 14},
    "Diana":  {"age": 50, "income": 85, "years_of_education": 18},
    "Edward": {"age": 29, "income": 44, "years_of_education": 13},
}
````

````python
# Answer:

# Part (a)

def total_distance(name1, name2, individuals):
    p1 = individuals[name1]
    p2 = individuals[name2]
    distance = abs(p1["age"] - p2["age"]) + abs(p1["income"] - p2["income"]) + abs(p1["years_of_education"] - p2["years_of_education"])
    return distance

# Part (b)

def find_match(name, individuals):
    best_match = None
    best_distance = None
    for other in individuals:
        if other != name:
            d = total_distance(name, other, individuals)
            if best_distance is None or d < best_distance:
                best_distance = d
                best_match = other
    return best_match, best_distance

# Part (c)

match, distance = find_match("Bob", individuals)
print(f"The best match for Bob is {match} with a total distance of {distance}.")
````

## Question 5: Designing a Household Survey (40 points)

In **development economics**, researchers often travel to developing countries to conduct **household surveys**. These surveys collect detailed information about individuals and households, including characteristics such as **age, income, education, employment status, caste/community, household size, and marital status**. The data collected from these surveys is used to study poverty, labor markets, education, inequality, and many other economic outcomes.

Traditionally, economists conduct these surveys **in person**, visiting villages and speaking directly with households. However, imagine a world where these questionnaires could be **completed online**, allowing individuals to enter their responses directly into a system.

In this question, you will build a **simple Python-based household survey questionnaire** that collects information from a respondent and stores it in a **dictionary**.

---

### Part A: Collecting Information

Write Python code that asks the user to input the following information:

- Full name  
- Place of residence (city or village)  
- Age  
- Education level  
- Employment status  
- Monthly income  
- Marital status  
- Household size  

You should use **`input()` statements** to collect this information.

---

### Part B: Data Validation

Some inputs must satisfy **basic validation checks**. Implement the following rules:

- **Age cannot be negative**
- **Monthly income cannot be negative**
- **Household size must be at least 1**

If the user enters an invalid value, your program should **ask the user to enter the value again**. You may use **`while` loops** to enforce these checks.

---

### Part C: Storing the Data

After collecting all responses, store the information in a **dictionary** called `household_survey`.

Each **key** should represent a survey question, and each **value** should store the user's response.

For example (structure only):

```python
household_survey = {
    "name": ...,
    "residence": ...,
    "age": ...,
    "education": ...,
    "employment_status": ...,
    "monthly_income": ...,
    "marital_status": ...,
    "household_size": ...
}

````python
# Answer: 

name = input("What is your full name? ")
residence = input("Where do you live (city or village)? ")

age = int(input("What is your age? "))
while age < 0:
    age = int(input("Age cannot be negative. Please enter your age again: "))

education = input("What is your education level? ")
employment_status = input("What is your employment status? ")

monthly_income = float(input("What is your monthly income? "))
while monthly_income < 0:
    monthly_income = float(input("Income cannot be negative. Please enter your monthly income again: "))

marital_status = input("What is your marital status? ")

household_size = int(input("What is your household size? "))
while household_size < 1:
    household_size = int(input("Household size must be at least 1. Please enter again: "))

household_survey = {
    "name": name,
    "residence": residence,
    "age": age,
    "education": education,
    "employment_status": employment_status,
    "monthly_income": monthly_income,
    "marital_status": marital_status,
    "household_size": household_size
}

print(household_survey)
````
