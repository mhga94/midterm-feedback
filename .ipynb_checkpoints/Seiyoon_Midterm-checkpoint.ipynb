# AI Governance Comparison Tool

This project uses OECD AI policy initiative data to compare AI governance across 10 countries. It groups policy initiatives into five categories — governance, regulation, ethics, capacity, and oversight — and then calculates a score for each country. The user can choose two countries to compare and receive simple recommendations based on policy gaps.

**The project is also turned into a Streamlit app via the app.py file inside the same directory folder.**

### Map OECD initiative types into 5 policy categories

#### Governance

These are broad state-planning and coordination efforts.

#### Regulation

These are harder rules, legal changes, or formal standards.

#### Ethics

These are softer guidance and norm-setting efforts.

#### Oversight

These involve review, accountability, monitoring, or institutional checking.

#### Capacity

These build infrastructure, funding, talent, compute, data, or diffusion.

```python
df_10 = df_10.copy()
```

```python
type_to_category = {
    "National AI strategy": "governance",
    "Guidance document (instructions on how to implement a law, regulation, policy or other rule)": "governance",
    "Key reports and whitepapers": "governance",
    "International and multi-stakeholder co-operation on AI": "governance",
    "Trade agreement that include AI": "governance",

    "Law/legislation/act (by legislative body)": "regulation",
    "Regulation (by government authority)": "regulation",
    "Amendment to/repeal of existing legislation": "regulation",
    "National Standard": "regulation",
    "Regulatory experimentation initiative including sandboxes": "regulation",

    "Principles/guidelines/frameworks for trustworthy AI": "ethics",
    "Public awareness/communications campaigns": "ethics",
    "Public consultations": "ethics",
    "Expert consultations": "ethics",

    "Advisory bodies, offices or processes": "oversight",
    "Coordination bodies, offices or processes": "oversight",
    "Accountability tool": "oversight",
    "Monitoring the impact of AI in the labour market": "oversight",

    "Research grants/funding": "capacity",
    "AI Research Centres/Centres of Excellence/Expertise, specialised AI teams, labs": "capacity",
    "Initiative/Investment in AI compute capacity": "capacity",
    "Initiative to foster access to AI compute capacity": "capacity",
    "Initiative/Investment in open datasets": "capacity",
    "Initiative to promote data access and sharing": "capacity",
    "Support for AI diffusion in firms": "capacity",
    "Support to start-ups": "capacity",
    "AI capacity/skills programme (primary or secondary education)": "capacity",
    "AI capacity/skills programme (public service)": "capacity",
    "AI capacity/skills programme (society, the public)": "capacity",
    "Formal education programme in higher education (Degree and fellowship programmes)": "capacity",
    "Reskilling/upskilling programme": "capacity",
    "Business grants/tax incentives": "capacity",
    "Talent attraction and retention (incl. immigration policies)": "capacity",
    "Science and innovation challenges, prizes and awards": "capacity",
    "International AI research collaboration": "capacity",
    "Initiative/Investment in natural language processing": "capacity",
    "Initiative/Investment in open-source tools": "capacity"
}
```

```python
# create a category column

df_10["policy_category"] = df_10["type"].map(type_to_category)
```

```python
df_10[["country", "type", "policy_category"]].head(15)
```

```python
df_10["policy_category"].value_counts(dropna=False)
```

```python
# Handle missing categories

df_10["policy_category"] = df_10["policy_category"].fillna("other")
```

```python
df_10["policy_category"].value_counts()
```

```python
# Initialize dictionary

country_policies = {}
```

```python
# Loop through dataset

for index, row in df_10.iterrows():
    
    country = row["country"]
    category = row["policy_category"]
    
    # if country not in dictionary, initialize it
    if country not in country_policies:
        country_policies[country] = {
            "governance": 0,
            "regulation": 0,
            "ethics": 0,
            "capacity": 0,
            "oversight": 0
        }
    
    # only count known categories
    if category in country_policies[country]:
        country_policies[country][category] += 1
```

```python
# check result

for country in country_policies:
    print(country, country_policies[country])
```

### Building Governance Score & Classification

```python
def calculate_policy_score(policy_dict):
    
    score = (
        policy_dict["governance"] * 1 +
        policy_dict["regulation"] * 2 +
        policy_dict["ethics"] * 1 +
        policy_dict["capacity"] * 1 +
        policy_dict["oversight"] * 2
    )
    
    return score
```

```python
# test the function

calculate_policy_score(country_policies["United States"])
```

```python
# apply to all countries

country_scores = {}

for country in country_policies:
    score = calculate_policy_score(country_policies[country])
    country_scores[country] = score
```

```python
# print results

for country in country_scores:
    print(country, country_scores[country])
```

```python
# converting the scores into labels (classification)

def classify_score(score):
    
    if score < 30:
        return "Low governance"
    elif score <= 60:
        return "Moderate governance"
    else:
        return "Strong governance"
```

```python
# apply classification to each country

for country in country_scores:
    score = country_scores[country]
    level = classify_score(score)
    print(f"{country}: {score} → {level}")
```

### Build Country Comparison

```python
country1 = input("Enter first country: ").strip()
country2 = input("Enter second country: ").strip()
```

```python
if country1 not in country_scores or country2 not in country_scores:
    print("One or both countries not found.")
else:
    print("Comparing", country1, "and", country2)
```

```python
score1 = country_scores[country1]
score2 = country_scores[country2]

print(country1, "score:", score1)
print(country2, "score:", score2)
```

```python
# Compare overall strength

if score1 > score2:
    print(country1, "has stronger AI governance overall.")
elif score2 > score1:
    print(country2, "has stronger AI governance overall.")
else:
    print("Both countries have similar governance strength.")
```

```python
# Compare category-by-category

print("\nCategory comparison:")

for category in country_policies[country1]:
    
    val1 = country_policies[country1][category]
    val2 = country_policies[country2][category]
    
    print(category, ":", country1, "=", val1, "|", country2, "=", val2)
```

```python
# Identify weaknesses (this powers recommendations)

if score1 < score2:
    weaker = country1
    stronger = country2
else:
    weaker = country2
    stronger = country1
```

### Add Recommendations

```python
print("\nRecommendations for", weaker + ":")

for category in country_policies[weaker]:
    
    weak_val = country_policies[weaker][category]
    strong_val = country_policies[stronger][category]
    
    if weak_val < strong_val:
        
        if category == "regulation":
            print("- Strengthen legal and regulatory frameworks.")
        
        elif category == "oversight":
            print("- Improve monitoring and accountability mechanisms.")
        
        elif category == "capacity":
            print("- Invest more in AI talent, research, and infrastructure.")
        
        elif category == "governance":
            print("- Develop clearer national AI strategies and coordination.")
        
        elif category == "ethics":
            print("- Expand ethical guidelines and responsible AI practices.")
```
