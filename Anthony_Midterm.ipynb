**Q26.** Write a function `incident_summary(incidents_list)` that:

- Takes a list of `(name, score)` tuples (like the `incidents` list from Section 5)
- Returns a **dictionary** with three keys:
  - `"total_incidents"`: count of incidents
  - `"average_score"`: average severity (rounded to 2 decimal places)
  - `"highest_score"`: the maximum score value

Call the function on the `incidents` list and print the returned dictionary.

````python
def incident_summary(incidents_list):
    total_incidents = len(incidents_list)

    total = 0
    highest_score = 0

    for incident, score in incidents_list:
        total += score
        if score > highest_score:
            highest_score = score

    average_score = round(total / total_incidents, 2)

    return {
        "total_incidents": total_incidents,
        "average_score": average_score,
        "highest_score": highest_score
    }

result = incident_summary(incidents)
print(result)
````

**Q29.** The following list contains troop strength estimates — but some entries are strings 
(data transmission errors), not numbers:
```python
troop_estimates = [12000, 8500, "unknown", 24000, "N/A", 15000, 9000, "classified"]
```

Write a loop using `try/except` that:

- Attempts to add each value to a running total
- On failure, prints: `"Skipping non-numeric entry: [value]"`
- After the loop, prints the total of all valid estimates and the count of skipped entries

````python
troop_estimates = [12000, 8500, "unknown", 24000, "N/A", 15000, 9000, "classified"]

total = 0
skipped = 0

for entry in troop_estimates:
    try:
        total += entry
    except:
        print("Skipping non-numeric entry:", entry)
        skipped += 1

print("Total valid estimate:", total)
print("Skipped entries:", skipped)
````
