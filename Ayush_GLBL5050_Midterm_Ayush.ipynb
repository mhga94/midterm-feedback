## Q1: Praying for Rain*

You are a priest in 14th-century Murcia, Spain. The community believes that religious rituals can bring rain during droughts. Your influence in the community depends on whether your ritual appears to “cause” the rain.

Each drought season begins with no rain and a very low probability that it will rain tomorrow. As the drought continues, the probability that it will rain the next day gradually increases. Each day you must decide whether to:
1. Conduct the rainmaking ritual
2. Wait another day

If you wait, the drought continues and you receive updated information on the probability that it will rain the next day. If you perform the ritual, the game records how many days later it rains. Your religious influence depends on how convincing the timing appears:
* If rain occurs soon after the ritual, people believe your ritual worked and your influence increases.
* If rain occurs long after the ritual, people are less convinced.
* If it rains before you perform the ritual, the community doubts your abilities and your influence decreases.

At the end of each drought season, the rainfall outcome is revealed and your religiosity score is updated. You will then begin a new drought season, starting with your current religiosity level. Your goal is to maximize your religiosity score over multiple drought seasons and become the most influential religious leader in Murcia.

Your program should simulate repeated drought seasons. At the start of each season the program prints:
```
A new drought season begins.
Current religiosity: (value)
Day 1
Probability of rain tomorrow: (value)
Do you perform the ritual today? (yes/no)
```
If the user chooses no, the program advances to the next day and prints the updated probability. If the user chooses yes, the program simulates when rain actually occurs and prints:
```
You performed the ritual on day X.
Rain occurred Y days later.
Your religiosity changes by Z.
Current religiosity: (value)
```
A new drought season then begins. The simulation continues until the user chooses to stop. Your program should handle invalid inputs gracefully and ask the user again if necessary.

````python
#importing random package
import random

religiosity = 50  #starting religiosity score

#loop
while True:
    print("--- A new drought season begins. ---")
    print(f"Current religiosity: {religiosity}")

    day = 1
    rain_prob = 0.05  #starting probability of rain tomorrow
    ritual_performed = False

    while not ritual_performed:
        print(f"Day {day}")
        print(f"Probability of rain tomorrow: {rain_prob:.2f}")

        #check if rain happens before ritual
        if random.random() < rain_prob: #the higher rain_prob, more likely that random returns a value below it (i.e. it rains before ritual)
            print("It rained before you performed the ritual!")
            penalty = -10 #penalty when rain occurs before ritual
            religiosity += penalty
            print(f"Your religiosity changes by {penalty}.")
            print(f"Current religiosity: {religiosity}")
            break

        #ask user whether to perform ritual
        while True:
            choice = input("Do you perform the ritual today? (yes/no): ").strip().lower()
            if choice in ("yes", "no"):
                break
            print("Invalid input. Please enter 'yes' or 'no'.")

        if choice == "yes":
            ritual_performed = True
            #simulate how many days until rain after the ritual
            days_after = 0
            current_prob = rain_prob
            while True:
                days_after += 1
                current_prob = min(current_prob + 0.05, 1.0) #checking if the added probability is still below 1
                if random.random() < current_prob: #similar logic as above, higher current_prob => lesser days_after
                    break

            #calculate religiosity change based on delay (scores)
            if days_after == 1:
                change = 15
            elif days_after == 2:
                change = 10
            elif days_after <= 4:
                change = 5
            elif days_after <= 7:
                change = 0
            else:
                change = -5

            religiosity += change
            print(f"You performed the ritual on day {day}.")
            print(f"Rain occurred {days_after} day(s) later.")
            print(f"Your religiosity changes by {change}.")
            print(f"Current religiosity: {religiosity}")
        else: #for when ritual not conducted
            day += 1
            rain_prob = min(rain_prob + 0.05, 1.0)

    #asking if user wants to continue
    while True:
        again = input("Start a new drought season? (yes/no): ").strip().lower()
        if again in ("yes", "no"):
            break
        print("Invalid input. Please enter 'yes' or 'no'.")

    if again == "no":
        print(f"Final religiosity: {religiosity}")
        print("Thank you for serving the people of Murcia!")
        break
````

## Q3: Simple Splitwise

I am headed tomorrow for a trip with friends to Peru! One of the most tidious parts of a group trip has always been keeping track of expenses. In that regard, the app Splitwise has been a godsend--- it allows one to record all transactions, including who paid for what, and at the end of the trip simplifies debts such that balance amounts are resolved with the fewest number of transactions. This question is an attempt to recreate the functioning of that app. 

Your program should allow the user to repeatedly enter information about group transactions. For each transaction, the user should enter:
* the name of the person who paid
* the names of all people included in that expense
* the total amount paid

Assume that every transaction is split equally among all people involved in that transaction, including the person who paid. After each transaction, the program should ask the user whether they want to:
1. Enter another transaction
2. Settle debts

If the user chooses to enter another transaction, the program should keep collecting expenses. If the user chooses to settle debts, the program should calculate the final balances and print the simplest possible set of payments needed to settle everything.

````python
balances = {}  # net balance for each person

while True:
    #get payer
    payer = input("Who paid? ").strip()

    #get people involved
    people_input = input("Who was included (comma-separated, including the payer)? ").strip()
    for p in people_input.split(","):
        people = [p.strip()]
    if payer not in people:
        people.append(payer)

    #get amount
    while True:
        try:
            amount = float(input("Total amount paid: ").strip())
            if amount <= 0:
                print("Amount must be positive.")
                continue
            break
        except ValueError:
            print("Invalid amount. Please enter a number.")

    #split equally
    share = amount / len(people)

    #update balances: payer is owed (amount - their share), others owe their share
    balances[payer] = balances.get(payer, 0) + (amount - share)
    for person in people:
        if person != payer:
            balances[person] = balances.get(person, 0) - share

    #ask what to do next
    while True:
        action = input("\n1. Enter another transaction\n2. Settle debts\nChoose (1/2): ").strip()
        if action in ("1", "2"):
            break
        print("Invalid choice. Please enter 1 or 2.")

    if action == "2": #settle debts option chosen
        break

# settle debts
print("\n--- Settlement ---")

#separating into debtors (negative balance) and creditors (positive balance)
debtors = []
creditors = []
for person, balance in balances.items():
    if balance < -0.01:
        debtors.append([person, -balance])  #store as positive amount owed
    elif balance > 0.01:
        creditors.append([person, balance])

# sort for consistent output (descending order)
debtors.sort(key=lambda x: x[1], reverse=True)
creditors.sort(key=lambda x: x[1], reverse=True)

if not debtors:
    print("No debts to settle. Everyone is even!")
else:
    i, j = 0, 0 #sequentially settling
    while i < len(debtors) and j < len(creditors):
        payment = min(debtors[i][1], creditors[j][1])
        print(f"{debtors[i][0]} pays {creditors[j][0]}: ${payment:.2f}")
        debtors[i][1] -= payment
        creditors[j][1] -= payment
        if debtors[i][1] < 0.01:
            i += 1
        if creditors[j][1] < 0.01:
            j += 1
````
