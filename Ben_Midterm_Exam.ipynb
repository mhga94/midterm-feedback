## Problem 1: Grocery Budget Checker

A student is shopping for the week and wants to know whether they stayed within budget.

Write a program that:

 1.	Asks the user for their weekly grocery budget.

```python
budget = float(input("What's your weekly grocery budget? "))
```

2. Asks for the prices of 3 grocery items.

```python
item1 = float(input("What's the price of grocery item 1: "))
item2 = float(input("What's the price of grocery item 2: "))
item3 = float(input("What's the price of grocery item 3: "))
```

3. Calculates the total spent.

```python
total_spent = item1 + item2 + item3
```

4.	Prints:
	the total spent,
	the amount left over if under budget,
	or the amount over budget if they overspent.

```python
print(f"Total spent: ${total_spent}")

if total_spent < budget:
    under_budget = budget - total_spent
    print(f"You are under budget by ${under_budget}")

elif total_spent > budget:
    over_budget = total_spent - budget
    print(f"You are over budget by ${over_budget}")

else:
    print("You spent exactly your budget.")
```

Small Twist
    "Great budgeting!" if they are under budget by more than $10
    "Very close!" if they are within $2 of the budget

```python
print(f"Total spent: ${total_spent}")

if total_spent < budget:
    under_budget = budget - total_spent
    print(f"You are under budget by ${under_budget:.2f}")

    if under_budget > 10:
        print("Great budgeting!")

    elif under_budget <= 2:
        print("Very close!")

elif total_spent > budget:
    over_budget = total_spent - budget
    print(f"You are over budget by ${over_budget:.2f}")

else:
    print("You spent exactly your budget.")
```

## Problem 2: Study Tracker Menu

A student wants to track how many hours they studied over several days.

Write a program that:
1. Creates an empty list called study_hours

```python
study_hours = []
```

2. Uses a while loop to repeatedly ask the user to enter study hours for a day and 
Stops when the user types "done"

```python
study_hours = []
while True:
    entry = input("Enter study hours for the day (type 'done' when you finish): ")
    if entry == "done":
        break
    hours = float(entry)
    study_hours.append(hours)
```

```python
print("Study hours list:", study_hours)

total_hours = sum(study_hours)
print("Total study hours:", total_hours)

average_hours = total_hours / len(study_hours)
print("Average study hours:", average_hours)

highest_hours = max(study_hours)
print("Highest hours studied in a day:", highest_hours)
```

## Problem 4: Country Trade Summary with Dictionaries

An economist is comparing exports and imports for a few countries.

```python
trade_data = { "Ghana": {"exports": 18, "imports": 14}, "Kenya": {"exports": 12, "imports": 15}, "Brazil": {"exports": 25, "imports": 25}}
```

Write a program that loops through the dictionary and, for each country:
1. Calculates the trade balance as exports - imports

```python
def trade_balance(exports, imports):
    return exports - imports

for country, values in trade_data.items():
    exports = values["exports"]
    imports = values["imports"]
    balance = trade_balance(exports, imports)
    print(country, balance)
```

    2. Prints whether the country has:
	a trade surplus,
	a trade deficit,
	or balanced trade

```python
for country, values in trade_data.items():
    exports = values["exports"]
    imports = values["imports"]
    balance = exports - imports

    if balance > 0:
        print(country, "has a trade surplus")

    elif balance < 0:
        print(country, "has a trade deficit")

    else:
        print(country, "has balanced trade")
```

```python
max_surplus = 0
max_country = ""

for country, values in trade_data.items():
    exports = values["exports"]
    imports = values["imports"]
    balance = exports - imports

    if balance > max_surplus:
        max_surplus = balance
        max_country = country
print("Largest trade surplus:", max_country)
```

## Problem 5: Function-Based Movie Ticket Calculator

```python
def ticket_price(age, is_student):
    if age < 13:
        price = 8
    elif age >= 13 and age <= 64:
        price = 12
    else:
        price = 7
    if is_student == "yes":
        price = price - 2
    if price < 5:
        price = 5
    return price
```

```python
age = float(input("What's your age? "))
```

```python
is_student = input("Are you a student? (yes or no)")
```

```python
age = float(input("Enter your age: "))
is_student = input("Are you a student? (yes/no): ")
price = ticket_price(age, is_student)
```

```python
print("Your ticket price is:", price)
```

```python
prices = []

for i in range(3):
    age = int(input("Enter age: "))
    student = input("Are you a student? (yes/no): ")

    price = ticket_price(age, student)
    prices.append(price)

average_price = sum(prices) / len(prices)

print("Average ticket price:", average_price)
```
