### Problem 5 — Price Deviation Detection

A hallmark of **tacit collusion** is when firms' prices move in lockstep, deviating little from each other even without explicit coordination. Antitrust economists look for **suspicious price parallelism** as a plus factor.

You are given monthly airline ticket prices (in dollars) for three routes operated by two carriers over 12 months:

```python
carrier_alpha = [312, 315, 314, 318, 317, 316, 322, 321, 323, 319, 320, 325]
carrier_beta  = [310, 314, 315, 317, 318, 315, 321, 322, 322, 320, 319, 324]
```

Write a program using a **`for` loop with `range()`** that:
1. Iterates over all 12 months.
2. For each month, computes the **absolute price difference** between the two carriers.
3. Prints: `"Month X: Alpha=$Y, Beta=$Z, Difference=$D"`
4. Counts how many months had a price difference of **$3 or less** (suspicious parallelism threshold).
5. After the loop, computes and prints the **average monthly price difference**.
6. Prints a final assessment: if more than **8 of 12 months** show parallelism, print `"WARNING: Price parallelism detected. Refer for investigation."`; otherwise print `"Prices appear competitive."`

````python
# Problem 5 — Your solution here
carrier_alpha = [312, 315, 314, 318, 317, 316, 322, 321, 323, 319, 320, 325]
carrier_beta = [310, 314, 315, 317, 318, 315, 321, 322, 322, 320, 319, 324]

month = 0
total_difference = 0
parallel_months = 0

for alpha_price, beta_price in zip(carrier_alpha, carrier_beta):
    month += 1
    difference = abs(alpha_price - beta_price)
    total_difference += difference
    if difference <= 3:
        parallel_months += 1
    print(f"Month {month}: Alpha=${alpha_price}, Beta=${beta_price}, Difference=${difference}")
print (f"Months with price parallelism: {parallel_months} months,{round(parallel_months/12*100)}% of studied period")
average_difference = total_difference / len(carrier_alpha)
print("Average price difference is", round(average_difference))

print (" === Final assessment ===")
if parallel_months > 8:
    print ("WARNING: Price parallelism detected. Refer for investigation.")
else:
    print("Prices appear competitive.")
````

### Problem 8 — Upward Pricing Pressure (UPP) Function Library

The **Upward Pricing Pressure (UPP)** index is a tool used by merger economists (developed by Farrell & Shapiro, cited in the Merger Guidelines) to predict whether a merger will lead to price increases, without requiring full market definition.

The formula for the UPP on firm 1 after a merger with firm 2 is:

$$UPP_1 = D_{12} \times (P_2 - C_2) - E \times C_1$$

Where:
- $D_{12}$ = diversion ratio from firm 1 to firm 2 (share of firm 1's lost sales that go to firm 2)
- $P_2$ = price of firm 2's product
- $C_2$ = marginal cost of firm 2
- $E$ = efficiency credit (standard assumption: 0.10, i.e., 10%)
- $C_1$ = marginal cost of firm 1

If UPP > 0, the merger is likely to raise prices for firm 1's product.

Write the following **three functions**:

**Function 1: `calc_upp(d12, p2, c2, c1, efficiency=0.10)`**
- Computes and returns the UPP value.
- Raises a `ValueError` with a descriptive message if any price or cost is negative, or if `d12` is not between 0 and 1.

**Function 2: `interpret_upp(upp_value, firm_name)`**
- Takes a UPP value and a firm name.
- Returns a string: `"Merger likely raises [firm]'s prices (UPP = X.XX)"` if UPP > 0, otherwise `"No upward pricing pressure on [firm] (UPP = X.XX)"`. Round to 2 decimal places.

**Function 3: `full_upp_analysis(merger_name, d12, d21, p1, p2, c1, c2, efficiency=0.10)`**
- Runs UPP analysis for **both** sides of the merger (firm 1 and firm 2).
- Calls `calc_upp` and `interpret_upp` for each firm.
- Prints a formatted report for the named merger.
- Handles `ValueError` exceptions gracefully, printing the error message instead of crashing.

Test your functions with the **AB InBev / Grupo Modelo** merger:
- `merger_name = "AB InBev / Grupo Modelo"`
- `d12 = 0.35` (35% of AB InBev customers switch to Modelo)
- `d21 = 0.28`
- `p1 = 18.50` (AB InBev price per case)
- `p2 = 16.75` (Modelo price per case)
- `c1 = 9.20`, `c2 = 8.10`

````python
# Problem 8 — Your solution here
def calc_upp(d12, p2, c2, c1, efficiency=0.10):
    UPP = (d12 * p2) - (c2 - c1) * (1 - efficiency)
    return UPP

def interpret_upp(UPP, firm_name):
    if UPP > 0:
        return f"Merger likely raises {firm_name}'s prices (UPP = {UPP:.2f})"
    else:
        return f"No upward pricing pressure on {firm_name} (UPP = {UPP:.2f})"

def full_upp_analysis(merger_name, d12, d21, p1, p2, c1, c2, efficiency=0.10):
    upp1 = calc_upp(d12, p2, c2, c1, efficiency)
    result1 = interpret_upp(upp1, "Firm 1")
    
    upp2 = calc_upp(d21, p1, c1, c2, efficiency)
    result2 = interpret_upp(upp2, "Firm 2")
    
    print(f"=== UPP Analysis: {merger_name} ===")
    print(result1)
    print(result2)
````

### Problem 9 — Leniency Discount Calculator with Nested Logic

Under the DOJ leniency program, cooperating cartel members receive **sentence reductions** based on: (1) whether they are the **first** to apply, (2) the **completeness** of their cooperation (rated 1–10), and (3) whether they provide evidence against **foreign** conspirators (a national interest bonus).

Write a function `calculate_leniency(firm_name, applicant_number, cooperation_score, foreign_evidence)`:

- `applicant_number` (int): 1 = first applicant, 2 = second, 3+ = subsequent
- `cooperation_score` (int, 1–10): quality of cooperation
- `foreign_evidence` (bool): whether the firm provides evidence against foreign firms

Discount logic:
- First applicant: base discount of **100%** (full amnesty) if `cooperation_score >= 8`; **80%** if score is 5–7; **50%** if below 5.
- Second applicant: base discount of **60%**, **40%**, or **20%** by the same score brackets.
- Third+ applicant: base discount of **30%**, **15%**, or **5%**.
- Foreign evidence bonus: **+10 percentage points** added to the discount, capped at 100%.
- Raise a `ValueError` if `cooperation_score` is outside the 1–10 range.

The function should **return** the final discount percentage and **print** a formatted leniency award letter for the firm.

Test with: Teva Pharmaceuticals, applicant #2, cooperation score 7, foreign evidence = True.

````python
# Problem 9 — Your solution here
def calculate_leniency(firm_name, applicant_number, cooperation_score, foreign_evidence):    
    if applicant_number == 1:
        if cooperation_score >= 8:
            discount = 100
        elif cooperation_score >= 5:
            discount = 80
        else:
            discount = 50
            
    elif applicant_number == 2:
        if cooperation_score >= 8:
            discount = 60
        elif cooperation_score >= 5:
            discount = 40
        else:
            discount = 20
            
    else:  # applicant 3 and beyond
        if cooperation_score >= 8:
            discount = 30
        elif cooperation_score >= 5:
            discount = 15
        else:
            discount = 5
    
    if foreign_evidence:
        discount += 10
    
    if discount > 100:
        discount = 100
    
    print("=== DOJ LENIENCY AWARD LETTER ===")
    print(f"Firm: {firm_name}")
    print(f"Applicant Order: #{applicant_number}")
    print(f"Cooperation Score: {cooperation_score}/10")
    print(f"Foreign Evidence Provided: {foreign_evidence}")
    print(f"Final Penalty Reduction: {discount}%") 
````
