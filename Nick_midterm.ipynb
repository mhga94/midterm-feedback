### Phase 1: Object-Oriented Intelligence (Classes & Objects)
Define the physical and corporate reality of space in 2026.
* **Class:** `Satellite`
* **Attributes:** `name`, `owner` (e.g., **Shanghai Spacecom**, **PLA**, **MizarVision**), `is_private` (Boolean), and `altitude` (Int).
* **Binding Method:** `evaluate_dual_use()`. This method should check if a satellite is "private" but operating in a strategic LEO altitude; if so, return a alert about "Military-Civil Fusion" (MCF).
* **Topics:** *Classes, Defining classes, Binding methods, Initializing objects.*

```python
class Satellite:
    def __init__(self, name, owner, is_private, altitude): #self is placeholder
        self.name = name #str
        self.owner = owner #str
        self.is_private = is_private #bool
        self.altitude = altitude #int

    def dualeval(self):
        if self.is_private and self.altitude < 1000:
            return f"WARNING: {self.name} is a private asset in a strategic orbit. Consider monitoring for MCF use."
        else:
            return f"{self.name} is operating within standard parameters."

s1 = Satellite("Qianfan-18", "Shanghai Spacecom", True, 550)
s2 = Satellite("TPG1000Cs-Unit1", "PLA Strategic Support", False, 0)
s3 = Satellite("Mizar-Alpha", "MizarVision", True, 480)
s4 = Satellite("Tianqi-41", "Guodian Gaoke", True, 600)
s5 = Satellite("SC-19 Interceptor", "CASC", False, 320)
s6 = Satellite("G60-Polar-02", "Spacesail", True, 1160)


print(s4.dualeval())
```

### Phase 5: Orbital Decay (Functions & Recursion)
Model the aftermath of a kinetic strike using the **Kessler Syndrome** logic.
* **Function:** `trigger_cascade(debris_count)`.
* **Recursion (Light):** Write a recursive function `calculate_decay(altitude)` that reduces altitude by 10% each "cycle" until it hits `100km` (re-entry).
* **Namespaces:** Use a **global** variable `TOTAL_DEBRIS` and a **local** variable inside your function to show how Python separates the two.
* **Topics:** *Functions, Parameters, Namespaces, Introductory recursion.*

```python
TOTAL_DEBRIS = 500

def trigger_cascade(debris_count):
    collision = debris_count*debris_count
    
    global TOTAL_DEBRIS
    TOTAL_DEBRIS += collision

    print(f"WARNING! The Kessler Syndrome exhibits exponential catastrophic debris count growth.\n{collision} new fragments detected.\nTotal debris count: {TOTAL_DEBRIS}")

userdeb = int(input("Enter a number of fragments:"))

trigger_cascade(userdeb)
```

```python
def signal_decode(intercept):
    if intercept > 100:
        return "Decoder key Sigma"
    elif intercept >= 66:
        return "Decoder key Epsilon"
    elif intercept >= 33:
        return "Decoder key Omega"
    else:
        return "Unknown key"

signal_decode(55)
```

TypeError -> The function is looking for a numberical value (integer) to compare against the given parameters.
