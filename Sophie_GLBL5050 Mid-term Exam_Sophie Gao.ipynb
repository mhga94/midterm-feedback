# Problem 2 — Classify people's support urgency

## Concept focus
Conditionals (if / elif / else)

## Task
Use the people dataset to classify each person into a support category.

## What code you must write
Write Python code that loops through the people dataset and classifies each person into one of these categories:
    "Low Support Need"
    "Moderate Support Need"
    "High Support Need"
    "Urgent Support Need"
Your logic must use at least three of the following variables:
    stress_level
    sleep_hours
    social_support_level
    emergency_level
    overall_need_level
Print each person's name and category.

This problem builds on Problem 1 because you now use the same dataset to make decisions with conditionals.

```python
for person in people:
    stress = person["stress_level"]
    sleep = person["sleep_hours"]
    support = person["social_support_level"]
    emergency = person["emergency_level"]
    overall = person["overall_need_level"]

    if emergency >= 2 or overall >= 9:
        category = "Urgent Support Need"
    elif stress >= 7 or sleep <= 5 or support <= 2:
        category = "High Support Need"
    elif stress >= 4 or overall >= 5:
        category = "Moderate Support Need"
    else:
        category = "Low Support Need"
    
    print(person["name"], ":", category)

for person in people:
    person["category"] = category
    print(person)
```

# Problem 7 — Write scoring functions

## Concept focus
Functions with parameters and return values

## Task
Now create simple scoring logic for people and companies.

## What code you must write
Write:
    A function called calculate_person_need(person)
    It should take one person dictionary as input and return a numeric need score.
    A function called calculate_company_support(company)
    It should take one company dictionary as input and return a numeric support score.
Your scoring logic should use meaningful attributes from each dataset.

Then write code that:
    Applies the person scoring function to every person
    Applies the company scoring function to every company
    Prints the results

This problem builds on everything earlier: you already understand the data, and now you turn that understanding into reusable logic.

```python
# A function called calculate_person_need(person)
def calculate_person_need(person):
    stress = person["stress_level"]
    sleep = person["sleep_hours"]
    support = person["social_support_level"]
    emergency = person["emergency_level"]
    need_score = (stress * 2 + (8 - sleep) + (5 - support) + emergency * 3)
    return need_score

# A function called calculate_company_support(company)
privacy_values = []
for company in companies:
    if company["data_privacy"] is not None:
        privacy_values.append(company["data_privacy"])
avg_privacy = sum(privacy_values) / len(privacy_values)

def calculate_company_support(company):
    therapy = company["therapy_depth"]
    personalization = company["personalization_level"]
    privacy = company["data_privacy"]
    if privacy is None:
        privacy = avg_privacy
    crisis = company["crisis_support_exists"]
    support_score = (therapy * 2 + personalization + privacy + (5 if crisis else 0))
    return support_score

for person in people:
    person_need_score = calculate_person_need(person)
    print("Person need scores:",person["name"], ":", person_need_score)

for company in companies:
    company_support_score = calculate_company_support(company)
    print("Company support scores:",company["company_name"], ":", company_support_score)
```

# Problem 8 — Handle missing data and create a simple matching class

## Concept focus
try/except for error handling, basic classes with attributes and methods

## Task
Some company data is incomplete. Your code should handle this safely. Then you will create a simple class to represent a match result.

## What code you must write
### Part A: Error handling
Write code using try/except that safely handles missing or invalid values when scoring companies.
For example, your code should not crash if:
    price_level is None
    data_privacy is None
Instead, it should print a warning and continue.

### Part B: Simple class
Create a class called MatchResult with:
    an attribute for person_name
    an attribute for company_name
    an attribute for match_score
Add a method called display_match() that prints a readable message about the match.

Then:
    Create at least two MatchResult objects
    Call the method for each object

This final problem builds on the entire exam: you use the datasets, scoring ideas, and safe programming practices to represent a matching result in a clean object-oriented way.

```python
# Part A:
privacy_values = []
price_values = []

for company in companies:
    if company["data_privacy"] is not None:
        privacy_values.append(company["data_privacy"])
    if company["price_level"] is not None:
        price_values.append(company["price_level"])

avg_privacy = sum(privacy_values) / len(privacy_values)
avg_price = sum(price_values) / len(price_values)


## Data cleaning - Detect and replace missing data
for company in companies:
    if company["data_privacy"] is None:
        print("Missing data_privacy for", company["company_name"])
        company["data_privacy"] = avg_privacy
    if company["price_level"] is None:
        print("Missing price_level for", company["company_name"])
        company["price_level"] = avg_price

## Error handling
def calculate_company_support(company):
    try:
        therapy = company["therapy_depth"]
        personalization = company["personalization_level"]
        privacy = company["data_privacy"]
        crisis = company["crisis_support_exists"]
        score = therapy * 2 + personalization + privacy + (5 if crisis else 0)
        return score
    except TypeError:
        print("TypeError while scoring", company["company_name"])
        return 0
    except KeyError:
        print("Missing key while scoring", company["company_name"])
        return 0

# Part B:
for person in people:
    matches = []
    person_need_score = calculate_person_need(person)
    for company in companies:
        company_support_score = calculate_company_support(company)
        match_score = company_support_score - person_need_score
        if person["budget_level"] < company["price_level"]:
            match_score -= 5
        if person["emergency_level"] > 0 and company["crisis_support_exists"]:
            match_score += 5
        matches.append((company["company_name"], match_score))
    
    matches.sort(key=lambda x: x[1], reverse=True)

    print("\nTop 2 companies for", person["name"])
    for company_name, score in matches[:2]:
        print(person["name"], "is matched with", company_name, "with a match score of", score)
```
