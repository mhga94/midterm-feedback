## LLM feedback

```python
API_KEY = getpass.getpass(prompt = "Enter your Gemini API key")
```

```python
# --- Configuration ---
MODEL_ID = "gemini-2.5-flash"
client = genai.Client(api_key=API_KEY)

def safe_print(message):
    """Prints strings safely by replacing characters the terminal cannot display."""
    try:
        clean_message = str(message).encode('ascii', 'ignore').decode('ascii')
        print(clean_message)
    except Exception:
        print("Log: [Message contained non-printable characters]")

def extract_notebook_content(notebook_path):
    """Extracts content and immediately cleans non-ASCII characters."""
    try:
        with open(notebook_path, 'r', encoding='utf-8') as f:
            nb = nbformat.read(f, as_version=4)
        
        formatted_content = []
        for i, cell in enumerate(nb.cells):
            header = f"--- Cell {i} ({cell.cell_type.upper()}) ---"
            content = cell.source.encode('ascii', 'ignore').decode('ascii')
            formatted_content.append(f"{header}\n{content}")
        
        return "\n\n".join(formatted_content)
    except Exception as e:
        return f"Error reading notebook: {str(e)}"

def extract_score(text):
    """Regex to pull the numerical score from the LLM response."""
    match = re.search(r'(?:Score|score):\s*(\d+)|(\d+)\s*/\s*100', text)
    if match:
        return match.group(1) or match.group(2)
    return "N/A"

def generate_midterm_feedback(notebook_file, chat_log_data):
    safe_print(f"Evaluating Submission: {os.path.basename(notebook_file)}")
    student_code = extract_notebook_content(notebook_file)
    
    prompt_parts = [
        "Act as a Teaching Assistant for 'Introduction to Python for Geopolitics'.",
        "Grade the following submission based strictly on the provided syllabus rubric.",
        "Submission Notebook Content:",
        student_code,
        "\nStudent's Supplementary Materials (Chat Logs/Docs):"
    ]

    chat_items = chat_log_data if isinstance(chat_log_data, list) else [chat_log_data]
    for item in chat_items:
        if str(item).startswith("http"):
            prompt_parts.append(f"- Shared Link: {item}")
        else:
            try:
                chat_file = client.files.upload(file=item)
                prompt_parts.append(chat_file)
            except Exception:
                safe_print(f"Warning: Could not process item {item}.")

    rubric_instruction = """
    Evaluate the submission using this 100-point scale:

    1. **Completion, Correctness, and Breadth (40% Total)**:
        - Did they complete the mid-term (19%)? 
        - Is the code entirely correct (1%)?
        - Does it cover a broad range of topics from the syllabus (20%)? The syllabus includes: variables, user input, if/elif/else, for loops, while loops, lists, dictionaries, list methods, string methods, functions, try/except error handling and basic imports.

    2. **LLM Interaction (40% Total)**:
        - Evaluate the chat log for thoughtful and logical interactions.
        - Did they refine prompts or explore tangents?
        - Did they show inquisitiveness about the applications of their learning?
    
    3. **Creativity and Novelty (20% Total)**:
        - Evaluate the subject matter choice and structure.
        - Look for 'amusing factoids' or long-form creative assignments.
        - Did they 'stretch' their skills (e.g., using try-except for incomplete data or new datasets)?

    Provide a final score out of 100 (formatted as 'Score: XX') and a 'Feedback for Student' section highlighting their strongest 'creative' element.
    """
    prompt_parts.append(rubric_instruction)

    response = client.models.generate_content(
        model=MODEL_ID,
        contents=prompt_parts,
        config=types.GenerateContentConfig(temperature=0.3)
    )

    return response.text

if __name__ == "__main__":
    output_csv = "midterm_feedback_results.csv"
    results = []

    for student_name, notebook_path in submissions_code.items():
        chat_data = submissions_chat.get(student_name, "No materials provided")
        
        try:
            # 1. Get the full response text from the LLM
            full_feedback = generate_midterm_feedback(notebook_path, chat_data)
            
            # 2. Parse out the score for a dedicated column
            numeric_score = extract_score(full_feedback)
            
            # 3. Explicitly build the dictionary for the CSV
            submission_entry = {
                "Student Name": student_name,
                "Score": numeric_score,
                "Full Feedback": full_feedback,  # This ensures the entire text is saved
                "Notebook Path": notebook_path,
                "Chat Materials": str(chat_data)
            }
            
            results.append(submission_entry)
            safe_print(f"Completed: {student_name} (Score: {numeric_score})")

        except Exception as e:
            clean_err = str(e).encode('ascii', 'ignore').decode('ascii')
            safe_print(f"Error processing {student_name}: {clean_err}")

    # --- CSV Writing Section ---
    if results:
        # Define the headers explicitly to match our dictionary keys
        headers = ["Student Name", "Score", "Full Feedback", "Notebook Path", "Chat Materials"]
        
        with open(output_csv, 'w', newline='', encoding='utf-8') as f:
            writer = csv.DictWriter(f, fieldnames=headers)
            writer.writeheader()
            writer.writerows(results)

    safe_print(f"Finished! CSV report generated: {output_csv}")
```
